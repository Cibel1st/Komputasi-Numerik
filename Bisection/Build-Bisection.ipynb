```python
import numpy as np

# Mendefinisikan fungsi non-linear yang akan dicari akarnya
# Contoh kasus: f(x) = x^3 - x - 2
def f(x):
    return x**3 - x - 2
```

```python
def metode_bisection(f, a, b, tol=1e-5, max_iter=100):
    # Mengecek syarat awal: f(a) dan f(b) harus memiliki tanda yang berbeda
    if f(a) * f(b) >= 0:
        print("Error: f(a) dan f(b) harus beda tanda (tidak ada akar di interval ini)!")
        return None, 0

    iterasi = 0
    c = a

    # Looping utama metode bisection
    while iterasi < max_iter:
        # Menghitung titik tengah interval
        c = (a + b) / 2
        f_c = f(c)

        # Menampilkan detail per iterasi
        print(f"Iterasi {iterasi+1}: a={a:.5f}, b={b:.5f}, c={c:.5f}, f(c)={f_c:.5f}")

        # Mengecek kriteria berhenti (konvergensi)
        if abs(f_c) < tol or (b - a) / 2 < tol:
            print(f"-> Konvergen pada iterasi ke-{iterasi+1}")
            break

        # Menentukan interval baru berdasarkan tanda f(c)
        if f(a) * f_c < 0:
            b = c # Akar berada di interval kiri
        else:
            a = c # Akar berada di interval kanan

        iterasi += 1

    return c, iterasi + 1
```

```python
# Menentukan parameter batas dan toleransi
# Untuk f(x) = x^3 - x - 2, akar berada di antara 1 dan 2
batas_bawah = 1
batas_atas = 2
toleransi = 1e-5

# Memanggil fungsi bisection
akar, total_iterasi = metode_bisection(f, batas_bawah, batas_atas, toleransi)

# Mencetak hasil akhir
if akar is not None:
    print(f"\nAkar persamaan: {akar:.6f}")
    print(f"Total iterasi: {total_iterasi}")
```

```python
# Memastikan nilai f(akar) sangat mendekati 0
print("Verifikasi nilai f(akar):", f(akar))
# Output yang diharapkan: angka sangat kecil (misal 1.2e-06)
```

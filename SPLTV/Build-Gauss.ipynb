```python
import numpy as np

# Sistem persamaan linear 3 variabel
# Contoh: 2x + y - z = 8 ; -3x - y + 2z = -11 ; -2x + y + 2z = -3
A = np.array([
    [ 2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
], dtype=float)
b = np.array([8, -11, -3], dtype=float)
```

```python
# Gabung A dan b jadi matriks augmented 3x4
Aug = np.hstack((A, b.reshape(-1, 1)))
print("Matriks augmented awal:")
print(Aug)
```

```python
def gauss_elimination(M):
    M = M.copy()
    n = M.shape[0]
    for k in range(n - 1):
        # Partial pivoting: pilih pivot terbesar di kolom k
        pivot_row = np.argmax(np.abs(M[k:, k])) + k
        if abs(M[pivot_row, k]) < 1e-12:
            raise ValueError("Sistem singular / tidak punya solusi unik")
        # Tukar baris
        if pivot_row != k:
            M[[k, pivot_row]] = M[[pivot_row, k]]
        # Eliminasi baris di bawah pivot
        for i in range(k + 1, n):
            factor = M[i, k] / M[k, k]
            M[i, k:] -= factor * M[k, k:]
    return M

Aug_elim = gauss_elimination(Aug)
print("Matriks setelah eliminasi:")
print(Aug_elim)
```

```python
def back_substitution(M):
    n = M.shape[0]
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (M[i, -1] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]
    return x

x = back_substitution(Aug_elim)
print("Solusi x1, x2, x3 =", x)
```
